```python
class student:
  def __init__(self,name, job):
    self.name=name
    self.job=job
std1= student("Rohit","Mumbai")
print(std1.name)
print(std1.job)
```

```python
class data:
    def __init__(self,name,age=19):
        self.name=name
        self.age=age
p1=data("Amit")
p2=data("Hitesh")
print(p1.name,p2.name,p2.age)
p2.age= 25
print(p1.name,p2.name,p2.age)
```

```python
class alldata:
    def __init__(self,name,age,city,hobby,married):
        self.name=name
        self.age=age
        self.city=city
        self.hobby=hobby
        self.married=married

person1= alldata("Farhan",21,"Cricket","Reading","Single")
print(person1.name,
person1.age,
person1.city,
person1.hobby,
person1.married)
```

```python
class farhan:
    def __init__(self,name):
        self.name=name
    def hell(self):
        print("Greeting",self.name)
to_greet= farhan("Warish")
print(to_greet.name)
to_greet.hell()
```

```python
class fullname:
    def __init__(self,fst,lst):
        self.fst=fst
        self.lst=lst
    def full(self):
        print(self.fst,self.lst)
call_name= fullname("Preet","pal")
call_name.full()
```
